## Introducción a Pandas

### ¿Qué es Pandas?

**Pandas** es la biblioteca más importante para análisis y manipulación de datos estructurados en Python. Construida sobre NumPy, proporciona estructuras de datos de alto nivel (Series y DataFrame) diseñadas específicamente para trabajar con datos tabulares y series temporales.

### ¿Por qué Pandas?

Si NumPy es como una calculadora científica potente, Pandas es como una hoja de cálculo (Excel) programable con superpoderes. Mientras NumPy trabaja con arrays numéricos homogéneos, Pandas maneja:

- **Datos heterogéneos:** Diferentes tipos en cada columna (números, texto, fechas)
- **Etiquetas significativas:** Índices con nombres descriptivos en lugar de solo números
- **Datos faltantes:** Manejo robusto de valores nulos (NaN)
- **Operaciones SQL-like:** Filtrado, agrupamiento, joins de forma intuitiva

### ¿Cuándo usar Pandas?

- **Análisis exploratorio de datos (EDA):** Entender patrones, estadísticas, distribuciones
- **Limpieza de datos:** El 80% del trabajo en Data Science
- **Transformación de datos:** Preparar datos para machine learning
- **Series temporales:** Datos con marcas de tiempo (finanzas, IoT, logs)
- **Reporting y visualización:** Generar tablas y gráficos para presentaciones

### Pandas vs NumPy

| Aspecto | NumPy | Pandas |
|---------|-------|--------|
| **Estructura principal** | ndarray (homogéneo) | DataFrame (heterogéneo) |
| **Índices** | Solo numéricos (0, 1, 2...) | Etiquetas personalizadas |
| **Datos faltantes** | Manejo básico | Manejo robusto (NaN, None) |
| **Velocidad** | Más rápido (operaciones numéricas) | Ligeramente más lento (más flexible) |
| **Uso típico** | Computación numérica, álgebra lineal | Análisis de datos, ETL |

**Conclusión:** Usa NumPy para cálculos numéricos intensivos, Pandas para análisis de datos y limpieza.

## 1. Series de Pandas

### ¿Qué es una Series?

Una **Series** es una estructura de datos unidimensional similar a un array de NumPy, pero con **superpoderes**.

**Analogía del mundo real - Lista de compras mejorada:**

**Array de NumPy (lista básica):**
```
[2.5, 3.0, 1.5, 4.0]
```
- ¿Qué significa cada número? No lo sabes sin consultar otra fuente
- Solo puedes acceder por posición: elemento[0], elemento[1]...

**Series de Pandas (lista con etiquetas):**
```
Manzanas 2.5 kg
Naranjas 3.0 kg
Plátanos 1.5 kg
Peras 4.0 kg
```
- Cada valor tiene una **etiqueta descriptiva**
- Puedes acceder por nombre: `compras['Manzanas']`
- Mucho más intuitivo y menos propenso a errores

### Diferencias clave entre NumPy Array y Pandas Series

| Característica | NumPy Array | Pandas Series |
|----------------|-------------|---------------|
| **Índice** | Solo numérico (0, 1, 2...) | Etiquetas personalizadas |
| **Acceso** | Solo por posición | Por etiqueta O por posición |
| **Nombre** | No tiene | Puede tener nombre descriptivo |
| **Tipos mixtos** | Mejor evitar | Más flexible (pero mejor homogéneo) |
| **Métodos** | Matemáticos/científicos | Análisis de datos + estadísticos |
| **Valores nulos** | Manejo básico | Manejo robusto (NaN) |

### Anatomía de una Series

```
       ÍNDICE VALORES NOMBRE
    ┌─────────────┐ ┌──────────┐ ↓
    │ Lunes │ → │ 23.5 │ "Temperatura (°C)"
    │ Martes │ → │ 25.0 │
    │ Miércoles │ → │ 22.8 │
    │ Jueves │ → │ 24.3 │
    │ Viernes │ → │ 26.1 │
    └─────────────┘ └──────────┘
         ↑ ↑
    (etiquetas) (datos)
```

**Componentes:**
1. **Valores (values):** Los datos en sí (array de NumPy internamente)
2. **Índice (index):** Etiquetas para cada valor
3. **Nombre (name):** Descripción opcional de qué representa la Series

### ¿Cuándo usar Series?

✓ **Usa Series cuando:**
- Tienes una secuencia de valores con significado (temperaturas, precios, puntuaciones)
- Quieres etiquetar cada valor con un nombre descriptivo
- Necesitas operar sobre datos 1D con estadísticas

✗ **No uses Series cuando:**
- Necesitas múltiples columnas → usa DataFrame
- Solo haces cálculos numéricos puros → NumPy es más rápido
- Los datos son realmente multidimensionales (matrices, imágenes)

### Crear Series

````python
# Forma 1: Desde una lista (índice automático 0, 1, 2...)
temperaturas = pd.Series([23.5, 25.0, 22.8, 24.3, 26.1])
print("Series con índice automático:")
print(temperaturas)
print(f"Tipo: {type(temperaturas)}")
print()
````

````python
# Forma 2: Con índice personalizado
temperaturas_dias = pd.Series(
    [23.5, 25.0, 22.8, 24.3, 26.1],
    index=['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes'],
    name='Temperatura (°C)'
)

print("Series con índice personalizado:")
print(temperaturas_dias)
print()

# Atributos importantes
print(f"Valores: {temperaturas_dias.values}") # Array de NumPy
print(f"Índice: {temperaturas_dias.index}") # Objeto Index
print(f"Nombre: {temperaturas_dias.name}")
print(f"Tipo de datos: {temperaturas_dias.dtype}")
````

````python
# Forma 3: Desde un diccionario (las claves se convierten en índice)
poblacion_ciudades = pd.Series({
    'Madrid': 3_223_334,
    'Barcelona': 1_636_762,
    'Valencia': 791_413,
    'Sevilla': 688_711,
    'Zaragoza': 674_997
})

print("Series desde diccionario:")
print(poblacion_ciudades)
print()
print(f"Shape: {poblacion_ciudades.shape}")
print(f"Tamaño: {poblacion_ciudades.size}")
````

### Indexación en Series

````python
# Acceso por etiqueta: siempre con .loc[]
print(f"Temperatura del Martes: {temperaturas_dias.loc['Martes']} C")
print(f"Poblacion de Barcelona: {poblacion_ciudades.loc['Barcelona']:,}")
print()

# Acceso por posicion: siempre con .iloc[]
print(f"Primera temperatura: {temperaturas_dias.iloc[0]} C")
print(f"Segunda ciudad: {poblacion_ciudades.iloc[1]:,}")
print()

# Por que .loc y .iloc y no los corchetes a secas.
#
# Una Series tiene etiquetas y tiene posiciones. Si el indice fuera
# numerico, serie[0] seria ambiguo: puede significar "la etiqueta 0" o
# "el primer elemento", y esas dos cosas no tienen por que coincidir.
# Pandas resolvia esa ambiguedad adivinando; a partir de la version 3
# ya no adivina y serie[0] con indice de etiquetas lanza un KeyError.
#
# La regla es corta: .loc para etiquetas, .iloc para posiciones, y los
# corchetes a secas solo para seleccionar columnas de un DataFrame.

# Recorte por etiquetas: INCLUYE el ultimo elemento.
print("De Lunes a Miércoles (por etiqueta, con .loc):")
print(temperaturas_dias.loc['Lunes':'Miércoles'])
print()

# Recorte por posicion: EXCLUYE el ultimo, como en Python.
print("Las tres primeras (por posicion, con .iloc):")
print(temperaturas_dias.iloc[0:3])
````

### Operaciones con Series

````python
# Operaciones aritméticas (vectorizadas como NumPy)
print("Convertir temperaturas a Fahrenheit:")
temp_fahrenheit = temperaturas_dias * 9/5 + 32
print(temp_fahrenheit)
print()

# Estadísticas descriptivas
print("Estadísticas de temperaturas:")
print(f"Media: {temperaturas_dias.mean():.2f}°C")
print(f"Desviación estándar: {temperaturas_dias.std():.2f}°C")
print(f"Mínima: {temperaturas_dias.min():.2f}°C")
print(f"Máxima: {temperaturas_dias.max():.2f}°C")
print(f"Mediana: {temperaturas_dias.median():.2f}°C")
print()

# Método describe() - resumen estadístico completo
print("Resumen estadístico completo:")
print(temperaturas_dias.describe())
````

````python
# Filtrado con máscaras booleanas
print("Días con temperatura > 24°C:")
dias_calurosos = temperaturas_dias[temperaturas_dias > 24]
print(dias_calurosos)
print()

# Ciudades con población > 1 millón
print("Ciudades con más de 1 millón de habitantes:")
ciudades_grandes = poblacion_ciudades[poblacion_ciudades > 1_000_000]
print(ciudades_grandes)
````

### Valores Nulos en Series

````python
# Series con valores faltantes
ventas_mensuales = pd.Series(
    [1200, np.nan, 1500, 1350, np.nan, 1600],
    index=['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun'],
    name='Ventas'
)

print("Series con valores nulos:")
print(ventas_mensuales)
print()

# Detectar valores nulos
print("¿Tiene valores nulos?")
print(ventas_mensuales.isna()) # o .isnull()
print()

print(f"Número de valores nulos: {ventas_mensuales.isna().sum()}")
print()

# Eliminar valores nulos
print("Serie sin valores nulos:")
print(ventas_mensuales.dropna())
print()

# Rellenar valores nulos con un valor
print("Serie con nulos rellenados con 0:")
print(ventas_mensuales.fillna(0))
print()

# Rellenar con la media
print("Serie con nulos rellenados con la media:")
print(ventas_mensuales.fillna(ventas_mensuales.mean()))
````

## 2. DataFrames: Creación y Estructura

### ¿Qué es un DataFrame?

Un **DataFrame** es una tabla bidimensional con columnas de tipos potencialmente diferentes. Es **LA** estructura de datos más importante en análisis de datos.

**Analogías del mundo real:**

1. **Hoja de cálculo de Excel:**
   - Filas y columnas con nombres
   - Cada columna puede tener diferente tipo (números, texto, fechas)
   - Puedes ordenar, filtrar, calcular

2. **Tabla de base de datos SQL:**
   - Estructura relacional con filas (registros) y columnas (campos)
   - Puedes hacer queries, joins, agregaciones

3. **Matriz NumPy con etiquetas:**
   - Como una matriz 2D de NumPy
   - Pero cada columna tiene nombre
   - Y cada fila tiene índice personalizable

### Anatomía de un DataFrame

```
               ┌──────── COLUMNAS ────────┐
               │ │
        ┌──────┴────────┬────────┬────────┴─────┐
        │ Nombre │ Edad │ Departamento│
    ┌───┼───────────────┼────────┼──────────────┤
    │ 0 │ Ana García │ 28 │ Ventas │
Í │ 1 │ Luis Pérez │ 34 │ IT │
N │ 2 │ Carlos Ruiz │ 29 │ IT │
D │ 3 │ María López │ 32 │ RRHH │
I │ 4 │ Pedro Sánchez │ 45 │ Ventas │
C └───┴───────────────┴────────┴──────────────┘
E ↑ ↑ ↑
         │ │ │
      (texto) (entero) (texto)
    
    ¡Cada columna puede tener un tipo diferente!
```

**Componentes principales:**

1. **Columnas:** Cada columna es una Series
   - Tienen nombre (ej: "Edad", "Salario")
   - Mismo tipo dentro de la columna (pero diferentes entre columnas)
   - Accesibles por nombre: `df['Edad']`

2. **Índice (filas):** Etiquetas para cada fila
   - Por defecto: 0, 1, 2, 3...
   - Personalizables: IDs, fechas, nombres...
   - Accesibles con `.loc[]` o `.iloc[]`

3. **Valores:** Los datos en sí
   - Internamente: matriz de NumPy (o múltiples arrays)
   - Accesibles con `.values`

### DataFrame vs Series vs NumPy

```
NumPy Array (2D): Solo números, sin etiquetas
┌───┬───┬───┐
│ 1 │ 2 │ 3 │
│ 4 │ 5 │ 6 │
└───┴───┴───┘

Pandas Series (1D): Una columna con etiquetas
       Valor
Lun 23
Mar 25
Mié 22

Pandas DataFrame (2D): Tabla completa con etiquetas
         Col1 Col2 Col3
Fila1 1 "A" True
Fila2 2 "B" False
         ↑ ↑ ↑
       (int) (str) (bool) ← Tipos diferentes OK
```

### ¿Por qué DataFrame es tan poderoso?

**1. Heterogeneidad de tipos:**
```python
# Una misma fila puede tener:
nombre: "Ana" (string)
edad: 28 (int)
salario: 35000.50 (float)
activo: True (bool)
fecha_ingreso: 2020-01-15 (datetime)
```

**2. Etiquetas significativas:**
```python
# En lugar de:
datos[3][1] # ¿Qué es esto?

# Puedes hacer:
df.loc['Ana García', 'Salario'] # ¡Mucho más claro!
```

**3. Operaciones SQL-like:**
```python
# Filtrar:
df[df['Edad'] > 30]

# Agrupar:
df.groupby('Departamento')['Salario'].mean()

# Join:
pd.merge(df_clientes, df_pedidos, on='ClienteID')
```

**4. Manejo robusto de datos faltantes:**
```python
# NaN se maneja automáticamente
df['Salario'].mean() # Ignora NaN automáticamente
```

**5. Integración perfecta con otras librerías:**
```python
# Matplotlib para gráficos
df.plot()

# Scikit-learn para ML
model.fit(df[features], df['target'])

# SQL databases
df.to_sql('tabla', connection)
```

### ¿Cuándo usar DataFrame?

✓ **Usa DataFrame cuando:**
- Tienes datos tabulares (filas y columnas)
- Diferentes columnas tienen diferentes tipos
- Necesitas filtrar, agrupar, combinar datos
- Trabajas con datasets reales (CSV, Excel, SQL)
- Haces análisis exploratorio de datos (EDA)

✗ **No uses DataFrame cuando:**
- Solo tienes una dimensión → usa Series
- Solo necesitas cálculos numéricos puros → usa NumPy
- Los datos son imágenes o tensores → usa NumPy o TensorFlow

## 3. Indexación y Selección Avanzada

### La Confusión más Grande de Pandas: loc vs iloc

Esta es probablemente **la fuente #1 de errores** para principiantes en Pandas. Pero una vez que entiendes la diferencia, todo tiene sentido.

**Analogía del mundo real - Encontrar un libro en una biblioteca:**

**Método 1: Por nombre (loc)**
- "Quiero el libro 'Don Quijote' de la sección 'Clásicos'"
- Usas las **etiquetas** (nombres) para encontrarlo
- No importa en qué estantería física está

**Método 2: Por posición física (iloc)**
- "Quiero el libro en la estantería 5, fila 3, posición 7"
- Usas la **ubicación física** (números)
- No importa cómo se llama el libro

### Comparación Visual: loc vs iloc

```
DataFrame de ejemplo:

        Producto Precio Stock ← Nombres de columnas
      ┌──────────────────────────┐
PROD001│ Laptop 899 15 │ ← Índice (etiqueta)
PROD002│ Mouse 25 120 │
PROD003│ Teclado 75 45 │
PROD004│ Monitor 299 30 │
      └──────────────────────────┘
       ↑ ↑ ↑ ↑
     col 0 col 1 col 2 posiciones


.loc['PROD002', 'Precio']
   ↑ ↑
etiqueta etiqueta
de fila de columna
= 25 ✓


.iloc[1, 1]
  ↑ ↑
posición posición
fila columna
= 25 ✓
```

### Reglas de Oro

| Aspecto | .loc[] | .iloc[] |
|---------|--------|---------|
| **Usa** | Etiquetas (nombres) | Posiciones (números enteros) |
| **Ejemplo fila** | `'PROD002'`, `'Madrid'` | `0`, `1`, `2` |
| **Ejemplo columna** | `'Precio'`, `'Stock'` | `0`, `1`, `2` |
| **Slicing** | INCLUYE el final | EXCLUYE el final |
| **Ventaja** | Más legible, robusto | Más rápido para loops |
| **Cuándo usar** | **Siempre que puedas** | Solo cuando posición importa |

### Diferencia CRÍTICA en Slicing

** ESTO ES MUY IMPORTANTE:**

```python
# Con .loc[] (slicing por etiquetas)
df.loc['PROD001':'PROD003']
# INCLUYE 'PROD001', 'PROD002', 'PROD003' (3 filas)
# El final SÍ se incluye

# Con .iloc[] (slicing por posición)
df.iloc[0:3]
# INCLUYE posiciones 0, 1, 2 (3 filas)
# El final NO se incluye (como en Python estándar)
```

**Regla mnemotécnica:**
- **loc** = "**L**abels **O**n **C**losing" → INCLUYE el final
- **iloc** = "**I**ndex **Loc**ation" (como Python) → EXCLUYE el final

### ¿Por qué usar .loc[] o .iloc[] explícitamente?

**MALA práctica (ambigua):**
```python
df['PROD002'] # ¿Es fila o columna? ¡Confuso!
df[0] # ¿Es índice o posición?
```

**BUENA práctica (clara):**
```python
df.loc['PROD002'] # Claramente la fila con etiqueta 'PROD002'
df.loc[:, 'Precio'] # Claramente la columna 'Precio'
df.iloc[1] # Claramente la fila en posición 1
```

### Casos de Uso

**Usa .loc[] cuando:**
✓ Trabajas con datos reales con índices significativos
✓ Quieres código legible y autodocumentado
✓ El orden de las filas puede cambiar
✓ Filtras con máscaras booleanas

**Usa .iloc[] cuando:**
✓ Iteras por posiciones en un loop
✓ Seleccionas "las primeras N filas" o "las últimas N filas"
✓ Trabajas con datos sin índice personalizado
✓ Necesitas máximo rendimiento en operaciones masivas

### loc: Indexación por Etiquetas

````python
# Seleccionar una fila por etiqueta
print("Fila con índice 'PROD001':")
print(df_ventas.loc['PROD001'])
print()

# Seleccionar un valor específico: fila y columna
print(f"Precio del producto PROD003: {df_ventas.loc['PROD003', 'Precio']}")
print()

# Seleccionar múltiples filas
print("Productos PROD001 y PROD003:")
print(df_ventas.loc[['PROD001', 'PROD003']])
print()

# Seleccionar filas y columnas específicas
print("Precio y Stock de PROD002 y PROD004:")
print(df_ventas.loc[['PROD002', 'PROD004'], ['Precio', 'Stock']])
````

## 4. Limpieza de Datos

### La Dura Realidad de los Datos Reales

**"Los datos reales son como adolescentes: nunca hacen lo que esperas y siempre necesitan supervisión."**

En el mundo académico, trabajas con datasets perfectos:
```
ID Nombre Edad Salario
1 Ana 28 35000
2 Luis 34 48000
3 María 32 38000
```

En el mundo real, te encuentras con esto:
```
ID Nombre Edad Salario
1 Ana 28 35000
2 LUIS NaN 48000
1 Ana 28 35000 ← Duplicado
3 Maria 32 38K ← Tipo incorrecto
4 150 52000 ← Nombre vacío, edad imposible
5 Pedro Perez -5 NULL ← Edad negativa
6 Unknown 32 999999 ← Outlier sospechoso
```

### ¿Por qué los datos están "sucios"?

**Fuentes comunes de problemas:**

1. **Entrada manual de datos:** Errores tipográficos, inconsistencias
   - "Madrid" vs "madrid" vs "MADRID" vs "Madrd"
   
2. **Sistemas legacy:** Datos de sistemas antiguos con formatos raros
   - Fechas como "01/02/03" (¿día/mes/año o mes/día/año?)
   
3. **Integración de múltiples fuentes:** Cada sistema tiene su formato
   - Base de datos A: NULL para faltantes
   - CSV de Excel: celdas vacías
   - API externa: "N/A" o "Unknown"
   
4. **Datos incompletos:** Formularios opcionales, sensores que fallan
   - Usuarios que dejan campos en blanco
   - Sensores IoT que pierden conexión

5. **Errores de medición:** Valores imposibles o incorrectos
   - Edad = 150 años
   - Temperatura = -999°C (código de error del sensor)

### La Regla del 80/20 en Data Science

**80% del tiempo:** Limpiando y preparando datos
**20% del tiempo:** Aplicando algoritmos y visualizando

**Por qué:** Los algoritmos de Machine Learning son como cocineros profesionales:
- Necesitan ingredientes limpios y bien preparados
- "Basura entra, basura sale" (GIGO: Garbage In, Garbage Out)
- Un modelo perfecto con datos sucios = resultados inútiles

### Tipos de Problemas en Datos

```
┌─────────────────────────────────────────────┐
│ PROBLEMAS COMUNES EN DATASETS REALES │
├──────────────────────┬──────────────────────┤
│ VALORES FALTANTES │ NaN, None, NULL, │
│ │ "", "N/A", "Unknown" │
├──────────────────────┼──────────────────────┤
│ DUPLICADOS │ Filas repetidas │
│ │ (error o legítimo) │
├──────────────────────┼──────────────────────┤
│ TIPOS INCORRECTOS │ "42" en vez de 42 │
│ │ "01/02/2024" como │
│ │ texto, no fecha │
├──────────────────────┼──────────────────────┤
│ INCONSISTENCIAS │ "madrid" vs "Madrid" │
│ │ Espacios extra │
├──────────────────────┼──────────────────────┤
│ OUTLIERS │ Valores extremos │
│ │ Errores de medición │
├──────────────────────┼──────────────────────┤
│ FORMATOS RAROS │ "10K" en vez de 10000│
│ │ "$1,234.56" como │
│ │ texto │
└──────────────────────┴──────────────────────┘
```

### Estrategias de Limpieza

No hay una solución única. Depende del **contexto** y del **objetivo**:

**Ejemplo: Valor faltante en "Edad"**

| Estrategia | Cuándo usarla | Ventaja | Desventaja |
|------------|---------------|---------|------------|
| **Eliminar fila** | Dataset grande, pocos nulos | Simple, datos 100% completos | Pierdes información |
| **Rellenar con media** | Análisis estadístico | Mantiene distribución | Subestima varianza |
| **Rellenar con mediana** | Hay outliers | Robusta a outliers | Menos precisa |
| **Rellenar con moda** | Datos categóricos | Valor más común | Puede sesgar |
| **Forward fill** | Series temporales | Lógica temporal | Propaga errores |
| **Predicción (ML)** | Pocos nulos, críticos | Más precisa | Complejo, costoso |

### Pipeline de Limpieza Típico

```
DATOS CRUDOS
    ↓
1. INSPECCIÓN INICIAL
   - .head(), .info(), .describe()
   - Identificar problemas
    ↓
2. VALORES FALTANTES
   - Detectar: .isna(), .isnull()
   - Decidir: ¿eliminar o rellenar?
   - Aplicar: .dropna() o .fillna()
    ↓
3. DUPLICADOS
   - Detectar: .duplicated()
   - Eliminar: .drop_duplicates()
    ↓
4. TIPOS DE DATOS
   - Verificar: .dtypes
   - Convertir: .astype(), pd.to_datetime()
    ↓
5. INCONSISTENCIAS
   - Normalizar texto: .str.lower(), .str.strip()
   - Corregir categorías
    ↓
6. OUTLIERS
   - Detectar: estadísticas, visualización
   - Decidir: ¿eliminar, corregir o mantener?
    ↓
DATOS LIMPIOS
    ↓
ANÁLISIS Y MODELADO
```

### Manejo de Valores Nulos

````python
# Crear DataFrame con valores faltantes
df_sucio = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', None, 'María', 'Pedro'],
    'Edad': [28, np.nan, 29, 32, 45],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', None, 'Madrid'],
    'Salario': [35000, 48000, np.nan, 38000, 52000]
})

print("DataFrame con valores nulos:")
print(df_sucio)
print()
````

````python
# Detectar valores nulos
print("Máscara de valores nulos:")
print(df_sucio.isna()) # o .isnull()
print()

# Contar valores nulos por columna
print("Valores nulos por columna:")
print(df_sucio.isna().sum())
print()

# Porcentaje de valores nulos
print("Porcentaje de nulos por columna:")
print((df_sucio.isna().sum() / len(df_sucio)) * 100)
````

````python
# Estrategia 1: Eliminar filas con CUALQUIER valor nulo
df_sin_nulos = df_sucio.dropna()
print("DataFrame eliminando CUALQUIER fila con nulos:")
print(df_sin_nulos)
print(f"Filas restantes: {len(df_sin_nulos)} de {len(df_sucio)}")
print()

# Estrategia 2: Eliminar filas donde TODAS las columnas son nulas
df_sin_nulos_todos = df_sucio.dropna(how='all')
print("Filas donde TODOS los valores son nulos (en este caso, ninguna):")
print(df_sin_nulos_todos)
print()

# Estrategia 3: Eliminar columnas con valores nulos
df_sin_cols_nulas = df_sucio.dropna(axis=1)
print("DataFrame eliminando columnas con nulos:")
print(df_sin_cols_nulas)
````

````python
# Estrategia 4: Rellenar valores nulos

# Con un valor constante
df_filled_cte = df_sucio.fillna(0)
print("Rellenar con 0:")
print(df_filled_cte)
print()

# Con valores diferentes por columna
df_filled_dict = df_sucio.fillna({
    'Nombre': 'Desconocido',
    'Edad': df_sucio['Edad'].mean(), # Media de edad
    'Ciudad': 'No especificada',
    'Salario': df_sucio['Salario'].median() # Mediana de salario
})

print("Rellenar con valores específicos por columna:")
print(df_filled_dict)
print()

# Forward fill (rellenar con valor anterior)
df_ffill = df_sucio.ffill()
print("Forward fill (usar valor anterior):")
print(df_ffill)
print()

# Backward fill (rellenar con valor siguiente)
df_bfill = df_sucio.bfill()
print("Backward fill (usar valor siguiente):")
print(df_bfill)
````

### Detección y Eliminación de Duplicados

````python
# Crear DataFrame con duplicados
df_con_duplicados = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Ana', 'María', 'Luis'],
    'Edad': [28, 34, 28, 32, 34],
    'Ciudad': ['Madrid', 'Barcelona', 'Madrid', 'Valencia', 'Barcelona']
})

print("DataFrame con duplicados:")
print(df_con_duplicados)
print()
````

````python
# Detectar filas duplicadas
print("Máscara de filas duplicadas:")
print(df_con_duplicados.duplicated())
print()

print(f"Número de filas duplicadas: {df_con_duplicados.duplicated().sum()}")
print()

# Ver las filas duplicadas
print("Filas duplicadas:")
print(df_con_duplicados[df_con_duplicados.duplicated()])
print()

# Eliminar duplicados (mantiene la primera ocurrencia)
df_sin_duplicados = df_con_duplicados.drop_duplicates()
print("DataFrame sin duplicados:")
print(df_sin_duplicados)
print()

# Mantener la última ocurrencia en lugar de la primera
df_sin_dup_last = df_con_duplicados.drop_duplicates(keep='last')
print("Sin duplicados (mantener última ocurrencia):")
print(df_sin_dup_last)
````

````python
# Detectar duplicados basados en columnas específicas
print("Duplicados basados solo en 'Nombre':")
print(df_con_duplicados[df_con_duplicados.duplicated(subset=['Nombre'])])
print()

# Eliminar duplicados basados en columnas específicas
df_sin_dup_nombre = df_con_duplicados.drop_duplicates(subset=['Nombre'])
print("Sin duplicados en 'Nombre':")
print(df_sin_dup_nombre)
````

### Conversión de Tipos de Datos

````python
# Crear DataFrame con tipos incorrectos
df_tipos = pd.DataFrame({
    'ID': ['1', '2', '3', '4', '5'], # Debería ser int
    'Precio': ['10.5', '20.3', '15', '30.9', '25'], # Debería ser float
    'Fecha': ['2024-01-15', '2024-02-20', '2024-03-10', '2024-04-05', '2024-05-12'], # Debería ser datetime
    'Activo': ['True', 'False', 'True', 'True', 'False'] # Debería ser bool
})

print("DataFrame con tipos incorrectos:")
print(df_tipos)
print()
print("Tipos de datos actuales:")
print(df_tipos.dtypes)
print()
````

````python
# Convertir tipos de datos
df_tipos['ID'] = df_tipos['ID'].astype(int)
df_tipos['Precio'] = df_tipos['Precio'].astype(float)
df_tipos['Fecha'] = pd.to_datetime(df_tipos['Fecha'])
df_tipos['Activo'] = df_tipos['Activo'].map({'True': True, 'False': False})

print("DataFrame con tipos corregidos:")
print(df_tipos)
print()
print("Tipos de datos corregidos:")
print(df_tipos.dtypes)
print()

# Ahora podemos hacer operaciones correctas
print(f"Suma de IDs: {df_tipos['ID'].sum()}")
print(f"Precio promedio: {df_tipos['Precio'].mean():.2f}")
print(f"Fecha más reciente: {df_tipos['Fecha'].max()}")
print(f"Número de activos: {df_tipos['Activo'].sum()}")
````

### Limpieza de Texto

````python
# DataFrame con texto sucio
df_texto_sucio = pd.DataFrame({
    'Nombre': [' Ana ', 'LUIS', 'carlos', ' MARÍA'],
    'Email': ['ana@email.com', 'LUIS@EMAIL.COM', 'carlos@email.com', None]
})

print("DataFrame con texto sucio:")
print(df_texto_sucio)
print()

# Limpiar espacios en blanco
df_texto_sucio['Nombre'] = df_texto_sucio['Nombre'].str.strip()

# Convertir a minúsculas
df_texto_sucio['Nombre'] = df_texto_sucio['Nombre'].str.lower()
df_texto_sucio['Email'] = df_texto_sucio['Email'].str.lower()

# Capitalizar nombres (primera letra mayúscula)
df_texto_sucio['Nombre'] = df_texto_sucio['Nombre'].str.capitalize()

print("DataFrame con texto limpio:")
print(df_texto_sucio)
````

## 5. Operaciones Básicas con DataFrames

### Ordenamiento

````python
# Crear DataFrame de ejemplo
df_estudiantes = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Carlos', 'María', 'Pedro'],
    'Nota': [8.5, 7.2, 9.1, 6.8, 8.9],
    'Edad': [22, 24, 21, 23, 22]
})

print("DataFrame original:")
print(df_estudiantes)
print()
````

````python
# Ordenar por una columna (ascendente)
df_ordenado_nota = df_estudiantes.sort_values('Nota')
print("Ordenado por Nota (ascendente):")
print(df_ordenado_nota)
print()

# Ordenar descendente
df_ordenado_nota_desc = df_estudiantes.sort_values('Nota', ascending=False)
print("Ordenado por Nota (descendente):")
print(df_ordenado_nota_desc)
print()

# Ordenar por múltiples columnas
df_ordenado_multi = df_estudiantes.sort_values(['Edad', 'Nota'], ascending=[True, False])
print("Ordenado por Edad (asc) y luego Nota (desc):")
print(df_ordenado_multi)
````

### Agregaciones y Estadísticas

````python
# Estadísticas básicas
print(f"Nota promedio: {df_estudiantes['Nota'].mean():.2f}")
print(f"Nota máxima: {df_estudiantes['Nota'].max()}")
print(f"Nota mínima: {df_estudiantes['Nota'].min()}")
print(f"Desviación estándar: {df_estudiantes['Nota'].std():.2f}")
print()

# Agregaciones múltiples
print("Agregaciones múltiples en Nota:")
print(df_estudiantes['Nota'].agg(['mean', 'median', 'std', 'min', 'max']))
print()

# Contar valores únicos
print("Distribución de edades:")
print(df_estudiantes['Edad'].value_counts())
````

### Aplicar Funciones

````python
# Aplicar función a una columna
def clasificar_nota(nota: float) -> str:
    """Clasifica una nota en categorías."""
    if nota >= 9:
        return 'Sobresaliente'
    elif nota >= 7:
        return 'Notable'
    elif nota >= 5:
        return 'Aprobado'
    else:
        return 'Suspenso'

df_estudiantes['Calificación'] = df_estudiantes['Nota'].apply(clasificar_nota)

print("DataFrame con calificaciones:")
print(df_estudiantes)
print()

# Función lambda (para funciones simples)
df_estudiantes['Nota_100'] = df_estudiantes['Nota'].apply(lambda x: x * 10)

print("DataFrame con Nota sobre 100:")
print(df_estudiantes)
````

## Resumen de Conceptos Clave

### Estructuras de Datos

1. **Series:**
   - Array 1D con índice personalizable
   - Crear: `pd.Series(valores, index=etiquetas)`
   - Acceso: por etiqueta o posición
   - Operaciones: aritméticas, estadísticas, filtrado

2. **DataFrame:**
   - Tabla 2D con columnas de tipos diferentes
   - Crear: desde dict, lista de dicts, arrays NumPy
   - Seleccionar columnas: `df['columna']` o `df[['col1', 'col2']]`
   - Añadir columnas: `df['nueva'] = valores`

### Indexación y Selección

1. **`.loc[]` - Por etiquetas:**
   - `df.loc[fila, columna]`
   - Slicing INCLUYE el último elemento
   - Usa nombres de índice/columna

2. **`.iloc[]` - Por posición:**
   - `df.iloc[posición_fila, posición_columna]`
   - Slicing EXCLUYE el último elemento
   - Usa números (0, 1, 2...)

3. **Filtrado booleano:**
   - `df[df['columna'] > valor]`
   - Combinar: `&` (AND), `|` (OR), `~` (NOT)
   - `.isin()` para múltiples valores

### Limpieza de Datos

1. **Valores nulos:**
   - Detectar: `.isna()`, `.isnull()`
   - Eliminar: `.dropna()`
   - Rellenar: `.fillna(valor)`, `.fillna(method='ffill')`

2. **Duplicados:**
   - Detectar: `.duplicated()`
   - Eliminar: `.drop_duplicates()`
   - Por columnas: `subset=['col1', 'col2']`

3. **Tipos de datos:**
   - Convertir: `.astype(tipo)`
   - Fechas: `pd.to_datetime()`
   - Mapeo: `.map(diccionario)`

### Operaciones

1. **Ordenamiento:**
   - `.sort_values(columna, ascending=True/False)`
   - Múltiples columnas: lista de nombres

2. **Agregaciones:**
   - `.mean()`, `.sum()`, `.min()`, `.max()`, `.std()`
   - `.describe()` para resumen completo
   - `.agg(['func1', 'func2'])` para múltiples

3. **Aplicar funciones:**
   - `.apply(función)` a Series/DataFrame
   - Lambda: `.apply(lambda x: ...)`

### Entrada/Salida

1. **CSV:**
   - Leer: `pd.read_csv('archivo.csv')`
   - Escribir: `df.to_csv('archivo.csv', index=False)`
   - Opciones: `sep`, `encoding`, `decimal`

### Mejores Prácticas

- ✓ Siempre usa `.loc[]` o `.iloc[]` para indexación
- ✓ Verifica tipos de datos con `.info()` y `.dtypes`
- ✓ Usa `.copy()` para evitar modificar DataFrames originales
- ✓ Explora datos con `.head()`, `.tail()`, `.describe()`
- ✓ Limpia datos antes de analizar (nulos, duplicados, tipos)
- ✗ Evita `SettingWithCopyWarning` usando `.loc[]`
- ✗ No asumas que los datos están limpios
- ✗ No modifies el DataFrame original sin `.copy()`
